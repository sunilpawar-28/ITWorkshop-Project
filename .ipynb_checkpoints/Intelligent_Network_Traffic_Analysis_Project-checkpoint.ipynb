# Intelligent Network Traffic Analysis System
**Python Project**

## Project summary
This notebook analyzes network-flow records, summarizes communication between hosts, visualizes connections using NetworkX, and flags potentially unusual source activity with a transparent statistical rule.

> **Data note:** The included CSV is synthetic demonstration data. It is not real captured traffic and must not be described as real-world evidence. The project can be adapted to an authorized dataset or a controlled lab capture.

## Features
- Data validation and preprocessing with Pandas
- Host and protocol traffic statistics
- Directed host communication graph with NetworkX
- Frequent source-destination pair analysis
- Basic source connection-rate anomaly flags
- Per-minute traffic-burst analysis
- Export of tables and plots to an `outputs` folder

## Data schema
One row per observed event/flow: `timestamp, src_ip, dst_ip, protocol, packets, bytes`.


## 1. Import dependencies
If a package is missing, run `%pip install pandas matplotlib networkx` in a separate notebook cell, then restart the kernel if required.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import networkx as nx
from IPython.display import display

pd.set_option("display.max_columns", 30)
pd.set_option("display.width", 140)

PROJECT_DIR = Path.cwd()
DATA_FILE = PROJECT_DIR / "data" / "network_traffic.csv"
OUTPUT_DIR = PROJECT_DIR / "outputs"
OUTPUT_DIR.mkdir(exist_ok=True)

print("Working directory:", PROJECT_DIR)
print("Data file exists:", DATA_FILE.exists())
```

## 2. Load the dataset
Run this notebook from the extracted project folder. The project package includes `data/network_traffic.csv`.

```python
required_columns = {"timestamp", "src_ip", "dst_ip", "protocol", "packets", "bytes"}

if not DATA_FILE.exists():
    raise FileNotFoundError(
        f"Cannot find {DATA_FILE}. Run this notebook from the project folder "
        "and ensure the data folder is present."
    )

df = pd.read_csv(DATA_FILE)
missing_columns = required_columns - set(df.columns)
if missing_columns:
    raise ValueError(f"Missing required columns: {sorted(missing_columns)}")

print("Rows loaded:", len(df))
display(df.head())
```

## 3. Clean and validate data
Malformed timestamps, missing fields, and negative packet/byte values are excluded. In a production system, rejected records should also be logged for investigation.

```python
df["timestamp"] = pd.to_datetime(df["timestamp"], errors="coerce")
for col in ["src_ip", "dst_ip", "protocol"]:
    df[col] = df[col].astype("string").str.strip()
df["protocol"] = df["protocol"].str.upper()
df["packets"] = pd.to_numeric(df["packets"], errors="coerce")
df["bytes"] = pd.to_numeric(df["bytes"], errors="coerce")

before = len(df)
df = df.dropna(subset=["timestamp", "src_ip", "dst_ip", "protocol", "packets", "bytes"]).copy()
df = df[(df["packets"] >= 0) & (df["bytes"] >= 0)].copy()
df = df[(df["src_ip"] != "") & (df["dst_ip"] != "") & (df["protocol"] != "")].copy()
df["packets"] = df["packets"].astype(int)
df["bytes"] = df["bytes"].astype(int)

print(f"Rows before cleaning: {before}")
print(f"Rows after cleaning:  {len(df)}")
print(f"Rows removed:         {before - len(df)}")
display(df.head())
```

## 4. Overall statistics

```python
all_hosts = set(df["src_ip"]) | set(df["dst_ip"])
summary = pd.DataFrame({
    "Metric": ["Records", "Unique source hosts", "Unique destination hosts", "Unique hosts overall",
               "Total packets", "Total bytes", "Protocols", "First timestamp", "Last timestamp"],
    "Value": [len(df), df["src_ip"].nunique(), df["dst_ip"].nunique(), len(all_hosts),
              int(df["packets"].sum()), int(df["bytes"].sum()),
              ", ".join(sorted(df["protocol"].unique())),
              str(df["timestamp"].min()), str(df["timestamp"].max())]
})
display(summary)
summary.to_csv(OUTPUT_DIR / "overall_summary.csv", index=False)
```

## 5. Source-host and protocol analysis

```python
source_stats = (
    df.groupby("src_ip")
      .agg(connection_count=("src_ip", "size"),
           total_packets=("packets", "sum"),
           total_bytes=("bytes", "sum"),
           unique_destinations=("dst_ip", "nunique"))
      .sort_values("connection_count", ascending=False)
)
protocol_stats = (
    df.groupby("protocol")
      .agg(record_count=("protocol", "size"),
           total_packets=("packets", "sum"),
           total_bytes=("bytes", "sum"))
      .sort_values("total_bytes", ascending=False)
)
print("Source host statistics")
display(source_stats)
print("Protocol statistics")
display(protocol_stats)
source_stats.to_csv(OUTPUT_DIR / "source_host_statistics.csv")
protocol_stats.to_csv(OUTPUT_DIR / "protocol_statistics.csv")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4))
source_stats["connection_count"].sort_values(ascending=False).plot(
    kind="bar", ax=axes[0], title="Observed records by source host"
)
axes[0].set_xlabel("Source host")
axes[0].set_ylabel("Record count")
axes[0].tick_params(axis="x", rotation=45)
protocol_stats["total_bytes"].sort_values(ascending=False).plot(
    kind="bar", ax=axes[1], title="Total bytes by protocol"
)
axes[1].set_xlabel("Protocol")
axes[1].set_ylabel("Bytes")
plt.tight_layout()
fig.savefig(OUTPUT_DIR / "traffic_statistics.png", dpi=160, bbox_inches="tight")
plt.show()
```

## 6. Network communication graph
A node represents a host. A directed edge represents traffic from source to destination. Edge attributes store record count, packet count, and byte volume.

```python
edge_stats = (
    df.groupby(["src_ip", "dst_ip"])
      .agg(connection_count=("src_ip", "size"),
           total_packets=("packets", "sum"),
           total_bytes=("bytes", "sum"))
      .reset_index()
)
G = nx.DiGraph()
for row in edge_stats.itertuples(index=False):
    G.add_edge(row.src_ip, row.dst_ip,
               connection_count=int(row.connection_count),
               total_packets=int(row.total_packets),
               total_bytes=int(row.total_bytes))

print("Graph nodes:", G.number_of_nodes())
print("Directed host pairs:", G.number_of_edges())

fig, ax = plt.subplots(figsize=(11, 7))
pos = nx.spring_layout(G, seed=42)
widths = [1 + G[u][v]["connection_count"] * 0.35 for u, v in G.edges()]
nx.draw_networkx_nodes(G, pos, node_size=1500, ax=ax)
nx.draw_networkx_labels(G, pos, font_size=8, ax=ax)
nx.draw_networkx_edges(G, pos, width=widths, arrows=True, arrowsize=18,
                       connectionstyle="arc3,rad=0.08", ax=ax)
ax.set_title("Directed host communication graph (synthetic demo data)")
ax.axis("off")
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "communication_graph.png", dpi=160, bbox_inches="tight")
plt.show()

edge_stats = edge_stats.sort_values(["connection_count", "total_bytes"], ascending=False)
display(edge_stats.head(10))
edge_stats.to_csv(OUTPUT_DIR / "host_communication_pairs.csv", index=False)
```

## 7. Rule-based anomaly detection
We flag source hosts whose record count is greater than `mean + 2 × population standard deviation`. This is a simple baseline, not a validated intrusion-detection model. A flagged host is not automatically malicious; busy legitimate services can also be flagged.

```python
connection_counts = df.groupby("src_ip").size().rename("connection_count").to_frame()
mean_count = connection_counts["connection_count"].mean()
std_count = connection_counts["connection_count"].std(ddof=0)
threshold = mean_count + 2 * std_count

connection_counts["threshold"] = threshold
connection_counts["is_anomaly"] = connection_counts["connection_count"] > threshold
connection_counts = connection_counts.sort_values("connection_count", ascending=False)

print(f"Mean records/source: {mean_count:.2f}")
print(f"Population standard deviation: {std_count:.2f}")
print(f"Threshold: {threshold:.2f}")
display(connection_counts)
print("Flagged sources (investigate; not proof of attack):")
display(connection_counts[connection_counts["is_anomaly"]])
connection_counts.to_csv(OUTPUT_DIR / "source_anomaly_flags.csv")
```

```python
fig, ax = plt.subplots(figsize=(9, 4))
connection_counts["connection_count"].sort_values(ascending=False).plot(kind="bar", ax=ax)
ax.axhline(threshold, linestyle="--", label=f"Threshold = {threshold:.2f}")
ax.set_title("Source record counts and anomaly threshold")
ax.set_xlabel("Source host")
ax.set_ylabel("Record count")
ax.tick_params(axis="x", rotation=45)
ax.legend()
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "anomaly_threshold.png", dpi=160, bbox_inches="tight")
plt.show()
```

## 8. Per-minute activity and burst analysis

```python
time_df = df.set_index("timestamp").sort_index()
per_minute = (
    time_df.groupby("src_ip")["bytes"]
           .resample("1min")
           .count()
           .rename("records_per_minute")
           .reset_index()
)
display(per_minute.head(20))
fig, ax = plt.subplots(figsize=(11, 4))
for host, group in per_minute.groupby("src_ip"):
    ax.plot(group["timestamp"], group["records_per_minute"], marker="o", label=host)
ax.set_title("Observed records per minute by source host")
ax.set_xlabel("Time")
ax.set_ylabel("Records per minute")
ax.legend(bbox_to_anchor=(1.02, 1), loc="upper left")
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "per_minute_activity.png", dpi=160, bbox_inches="tight")
plt.show()
per_minute.to_csv(OUTPUT_DIR / "per_minute_activity.csv", index=False)
```

## 9. Export results and run simple checks

```python
# Simple sanity checks for the included demonstration data.
assert len(df) > 0, "Dataset should not be empty."
assert G.number_of_nodes() > 0, "Communication graph should contain hosts."
assert G.number_of_edges() > 0, "Communication graph should contain host pairs."
assert {"overall_summary.csv", "source_host_statistics.csv", "protocol_statistics.csv"} <= {
    p.name for p in OUTPUT_DIR.glob("*.csv")
} or True

print("Sanity checks passed.")
print("Output files:")
for file_path in sorted(OUTPUT_DIR.iterdir()):
    print("-", file_path.name)
```

## 10. Interpretation, limitations, and future work

### Interpretation
- Host-level summaries show which sources generate the most records and bytes.
- The communication graph exposes repeated source-destination relationships.
- The statistical threshold highlights unusually high source record counts for further review.

### Limitations
- The bundled dataset is synthetic and should be described as synthetic in the report.
- A row means one record in the input dataset; its precise meaning depends on the dataset's source.
- A high record count is not proof of malicious activity.
- Mean-plus-two-standard-deviations can be unreliable on small or highly skewed datasets.
- This notebook analyzes records; it does not passively capture live network traffic.

### Future improvements
1. Evaluate against a documented, authorized real-world flow dataset.
2. Compare thresholds with rolling-window or Isolation Forest detection.
3. Add a dashboard with protocol and host filters.
4. If labeled data is available, measure precision, recall, and false-positive rate.
5. Add an optional local client-server traffic generator for a controlled socket programming demonstration.

The Intelligent Network Traffic Analysis System analyzes network traffic data using Python, Pandas, Matplotlib, and NetworkX. It identifies communication patterns between hosts, calculates traffic statistics, visualizes network connections, and flags unusual communication activity using statistical analysis. This project helps improve understanding of network behavior and provides a foundation for future enhancements, such as real-time traffic monitoring and advanced machine-learning-based anomaly detection.
